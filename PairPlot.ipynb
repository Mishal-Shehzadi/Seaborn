```python
sns.pairplot(data = penguins[columns])
```

```python
sns.pairplot(data = penguins)
```

```python
sns.pairplot(data = penguins, hue = "sex", palette = "Set1", diag_kind="kde")
```

```python
sns.pairplot(data = penguins, hue = "sex", palette = "Set1", diag_kind="hist")
```
