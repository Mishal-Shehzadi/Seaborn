```python
import seaborn as sns
import pandas as pd
import numpy as np
penguins = sns.load_dataset('penguins')
```

```python
graph = sns.PairGrid (data = penguins, hue = "sex", palette= "Set1")
graph.map_upper(sns.scatterplot) #upper trinagle
graph.map_lower(sns.kdeplot) #lower trinagle
graph.map_diag(sns.histplot) #diagonal
graph.add_legend()
```
