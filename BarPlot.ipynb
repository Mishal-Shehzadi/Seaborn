```python
import seaborn as sns
import pandas as pd
import numpy as np
penguins = sns.load_dataset('penguins')
```

```python
 sns.barplot (data = penguins, x = "species", y = "flipper_length_mm", hue = "sex", palette = ["violet", "skyblue"])
```

```python
sns.barplot (data = penguins, x = "species", y = "flipper_length_mm", hue = "sex", palette = ["violet", "skyblue"], estimator = np.sum)
```

```python
sns.barplot (data = penguins, x = "species", y = "flipper_length_mm", hue = "sex", palette = ["violet", "skyblue"], estimator = np.median)
```

```python
sns.barplot (data = penguins, x = "species", y = "flipper_length_mm", hue = "sex", palette = ["violet", "skyblue"], estimator = np.std)
```

```python
sns.barplot (data = penguins, x = "species", y = "flipper_length_mm", hue = "sex", palette = ["violet", "skyblue"], estimator = np.var)
```
