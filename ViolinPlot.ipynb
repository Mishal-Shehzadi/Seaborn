```python
import seaborn as sns
import pandas as pd
import numpy as np
penguins = sns.load_dataset('penguins')
```

```python
 sns.violinplot (data = penguins, x = "species", y = "body_mass_g", hue = "island")
```

```python
 sns.violinplot (data = penguins, x = "species", y = "body_mass_g", hue = "sex", split = True)
```

```python
sns.violinplot (data = penguins, x = "species", y = "body_mass_g")
sns.swarmplot (data = penguins, x = "species", y = "body_mass_g", color = "black", size = 3)
```

```python
sns.violinplot (data = penguins, x = "species", y = "body_mass_g", palette = "pastel")
sns.swarmplot (data = penguins, x = "species", y = "body_mass_g", hue = "island", size = 3, palette = "dark")
```
