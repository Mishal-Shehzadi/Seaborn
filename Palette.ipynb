```python
import seaborn as sns
import pandas as pd
import numpy as np
penguins = sns.load_dataset('penguins')
```

```python
#set1
sns.set_style ("ticks")
sns.set_context("notebook")
sns.scatterplot(data = penguins, x = "flipper_length_mm", y = "body_mass_g", hue = "island", palette = "Set1" )
sns.despine(left = True)
```
